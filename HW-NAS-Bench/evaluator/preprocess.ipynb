# Preprocesses the dataset to separate the metrics for the different edge devices

```python
import pandas as pd
import numpy as np
```

```python
search_space = 'nasbench201'
```

```python
df = pd.read_csv(f'../{search_space}_cifar10.csv')
df.head()
```

```python
# edge_gpu_df = df[['Architectures', 'Accuracy', 'Memory', 'EdgeGPU_Lat']].copy() # NAS-Bench201
edge_gpu_df = df[['Architectures', 'EdgeGPU_Lat']].copy()  # FBNet
edge_gpu_df.rename(columns={'EdgeGPU_Lat': 'Latency'}, inplace=True)
edge_gpu_df.to_csv(f'profiled_nets/{search_space}_edgegpu.csv', index=False)
edge_gpu_df
```

```python
# FBNet does not have Edge TPU
edge_tpu_df = df[['Architectures', 'Accuracy', 'EdgeTPU_Lat']].copy()
edge_tpu_df.rename(columns={'EdgeTPU_Lat': 'Latency'}, inplace=True)
edge_tpu_df.to_csv(f'profiled_nets/{search_space}_edgetpu.csv', index=False)
edge_tpu_df.head()
```

```python
# raspi4_df = df[['Architectures', 'Accuracy','Memory', 'Raspi4_Lat']].copy()
raspi4_df = df[['Architectures', 'Raspi4_Lat']].copy()
raspi4_df.rename(columns={'Raspi4_Lat': 'Latency'}, inplace=True)
raspi4_df.to_csv(f'profiled_nets/{search_space}_raspi4.csv', index=False)
raspi4_df.head()
```

```python
# eyeriss_df = df[['Architectures', 'Accuracy', 'Memory', 'Eyeriss_Lat']].copy()
eyeriss_df = df[['Architectures', 'Eyeriss_Lat']].copy()
eyeriss_df.rename(columns={'Eyeriss_Lat': 'Latency'}, inplace=True)
eyeriss_df.to_csv(f'profiled_nets/{search_space}_eyeriss.csv', index=False)
eyeriss_df.head()
```

```python
# pixel3_df = df[['Architectures', 'Accuracy', 'Memory', 'Pixel3_Lat']].copy()
pixel3_df = df[['Architectures', 'Pixel3_Lat']].copy()
pixel3_df.rename(columns={'Pixel3_Lat': 'Latency'}, inplace=True)
pixel3_df.to_csv(f'profiled_nets/{search_space}_pixel3.csv', index=False)
pixel3_df.head()
```

```python
# fpga_df = df[['Architectures', 'Accuracy', 'Memory', 'FPGA_Lat']].copy()
fpga_df = df[['Architectures', 'FPGA_Lat']].copy()
fpga_df.rename(columns={'FPGA_Lat': 'Latency'}, inplace=True)
fpga_df.to_csv(f'profiled_nets/{search_space}_fpga.csv', index=False)
fpga_df.head()
```
